```python
import pandas as pd
from os import listdir
from os.path import isfile, join
import numpy as np

from pyprojroot import here 

wd = str(here('data/raw-data/College Scorecard/'))
pub = str(here('data/processed-data'))

files = [f for f in listdir(wd) if isfile(join(wd, f))]
```

```python
cols = ['UNITID','GRAD_DEBT_MDN','GRAD_DEBT_MDN_SUPP','GRAD_DEBT_MDN','MD_EARN_WNE_P6','MD_EARN_WNE_P8','MD_EARN_WNE_P10', 'file']
data = pd.DataFrame(columns = cols)

for i in range(len(files)):
    temp = pd.read_csv(wd+'\\'+files[i])[['UNITID','GRAD_DEBT_MDN','GRAD_DEBT_MDN_SUPP','GRAD_DEBT_MDN','MD_EARN_WNE_P6','MD_EARN_WNE_P8','MD_EARN_WNE_P10',]]
    temp['file'] = files[i]
    data = pd.concat([data,temp])

data = data.reset_index().drop(columns='index')
```

```python
# debt is calculated for pooled cohorts (two most recent per year). Calculate min and max using file name and format
data['min_cohort'] = data['file'].str.strip('MERGED').str.strip('_PP.csv').str.split('_',expand=True)[0].astype(int)
data['max_cohort'] = data['file'].str.strip('MERGED').str.strip('_PP.csv').str.split('_',expand=True)[0].astype(int)+1

data['min_cohort'] = data['min_cohort'].astype(str)
data['max_cohort'] = data['max_cohort'].astype(str).str[-2:]

data['cohorts'] = data['min_cohort'] + '-' + data['max_cohort']

data['data_year'] = data['min_cohort'].astype(int)+6
```

```python
data.to_csv(pub + '\\college_scorecard_median_debt_earnings.csv')
```
